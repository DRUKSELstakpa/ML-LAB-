```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.arange(1, 51)
noise = np.random.normal(0, 20, 50)
y = 4 * X + 15 + noise

n = len(X)

mean_x = np.mean(X)
mean_y = np.mean(y)

numerator = np.sum((X - mean_x) * (y - mean_y))
denominator = np.sum((X - mean_x) ** 2)

m = numerator / denominator
b = mean_y - m * mean_x

y_pred = m * X + b

mse = np.mean((y - y_pred) ** 2)
r2 = 1 - (np.sum((y - y_pred) ** 2) / np.sum((y - mean_y) ** 2))

print("Slope:", m)
print("Intercept:", b)
print("MSE:", mse)
print("R2 Score:", r2)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.show()
```
