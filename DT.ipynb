```python
import math

data = [
    [2, 0],
    [3, 0],
    [4, 0],
    [6, 1],
    [7, 1],
    [8, 1]
]

def entropy(rows):
    total = len(rows)
    if total == 0:
        return 0

    count0 = sum(1 for row in rows if row[1] == 0)
    count1 = sum(1 for row in rows if row[1] == 1)

    e = 0

    for count in [count0, count1]:
        if count != 0:
            p = count / total
            e -= p * math.log2(p)

    return e

parent_entropy = entropy(data)

best_gain = -1
best_split = None

for row in data:
    threshold = row[0]

    left = [r for r in data if r[0] < threshold]
    right = [r for r in data if r[0] >= threshold]

    n = len(data)

    weighted_entropy = (
        len(left) / n * entropy(left)
        + len(right) / n * entropy(right)
    )

    gain = parent_entropy - weighted_entropy

    if gain > best_gain:
        best_gain = gain
        best_split = threshold

def predict(x):
    if x < best_split:
        return 0
    else:
        return 1

test = 7

print("Best Split:", best_split)
print("Information Gain:", best_gain)
print("Prediction:", predict(test))
```
