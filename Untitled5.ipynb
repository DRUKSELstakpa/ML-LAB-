```python
import math
```

```python
def entropy(data):
    label_count = {}
    for row in data:
        label = row[-1]
        if label not in label_count:
            label_count[label] = 0
        label_count[label] += 1
    ent = 0
    total = len(data)
    for key in label_count:
        p = label_count[key] / total
        ent -= p * math.log2(p)
    return ent
```

```python
def split_data(data, index, value):
    subset = []
    for row in data:
        if row[index] == value:
            subset.append(row)
    return subset
```

```python
def information_gain(data, index):
    base_entropy = entropy(data)
    values = set([row[index] for row in data])
    weighted_entropy = 0
    total = len(data)
    for v in values:
        subset = split_data(data, index, v)
        weighted_entropy += (len(subset) / total) * entropy(subset)
    return base_entropy - weighted_entropy
```

```python
def best_feature(data):
    gains = []
    for i in range(len(data[0]) - 1):
        gains.append(information_gain(data, i))
    return gains.index(max(gains))
```

```python
def majority_label(data):
    count = {}
    for row in data:
        label = row[-1]
        if label not in count:
            count[label] = 0
        count[label] += 1
    return max(count, key=count.get)
```

```python
def build_tree(data, features):
    labels = [row[-1] for row in data]
    if labels.count(labels[0]) == len(labels):
        return labels[0]
    if len(features) == 0:
        return majority_label(data)
    best = best_feature(data)
    tree = {features[best]: {}}
    values = set([row[best] for row in data])
    for v in values:
        subset = split_data(data, best, v)
        new_features = features[:best] + features[best+1:]
        new_subset = [row[:best] + row[best+1:] for row in subset]
        tree[features[best]][v] = build_tree(new_subset, new_features)
    return tree
```

```python
data = [
    ['Sunny', 'Hot', 'High', 'Weak', 'No'],
    ['Sunny', 'Hot', 'High', 'Strong', 'No'],
    ['Overcast', 'Hot', 'High', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'High', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Strong', 'No'],
    ['Overcast', 'Cool', 'Normal', 'Strong', 'Yes']
]

features = ['Outlook', 'Temp', 'Humidity', 'Wind']

tree = build_tree(data, features)
print(tree)
```
