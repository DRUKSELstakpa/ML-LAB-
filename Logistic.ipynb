```python
import math

X = [1, 2, 3, 4, 5, 6]
Y = [0, 0, 0, 1, 1, 1]

w = 0
b = 0

learning_rate = 0.1
epochs = 1000

def sigmoid(z):
    return 1 / (1 + math.exp(-z))

for _ in range(epochs):
    dw = 0
    db = 0

    for x, y in zip(X, Y):
        z = w * x + b
        y_pred = sigmoid(z)

        dw += (y_pred - y) * x
        db += (y_pred - y)

    w -= learning_rate * dw / len(X)
    b -= learning_rate * db / len(X)

test = 4.5

probability = sigmoid(w * test + b)

prediction = 1 if probability >= 0.5 else 0

print("Probability:", probability)
print("Prediction:", prediction)
```
